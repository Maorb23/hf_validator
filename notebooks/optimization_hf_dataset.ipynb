# Prepare an immutable HF optimization workload
CPU preparation only. Install updated llm-optimizer plus datasets and transformers.
Use HF_TOKEN in the environment or cached HF authentication. Never enter credentials here.
First-human tokens are repeated/truncated to input_tokens; output_tokens is the requested output length.
This notebook does not submit jobs. Share the same file across frameworks and trials.

```python
import json
from pathlib import Path
from llm_optimizer.hf_workload import prepare_hf_workload

dataset = {"repo_id": "ORG/DATASET", "split": "train"}  # optional revision/config
model = "ORG/MODEL"
model_revision = "REPLACE_WITH_MODEL_COMMIT"
manifest = prepare_hf_workload(dataset, model, model_revision,
    input_tokens=512, output_tokens=128, num_prompts=100, seed=7)
path = Path("workload-" + manifest["sha256"] + ".json")
with path.open("x", encoding="utf-8") as target:
    json.dump(manifest, target, ensure_ascii=False)
print({"path": str(path.resolve()), "sha256": manifest["sha256"],
       "dataset": manifest["dataset"], "tokenizer_revision": manifest["tokenizer_revision"]})
```
